```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from utils import plot_bivariate_gaussian, plot_gaussian_mixture

savepath = Path(
    "/Users/shreyaspadhy/Documents/Obsidian Vault/phd-thesis/Chapter2/Figs/"
)
# Create the figure
fig = plt.figure(figsize=(5, 5))

# First subplot: rho = 0.9
ax1 = fig.add_subplot(1, 1, 1, projection="3d")
plot_bivariate_gaussian(ax1, rho=0.0, title=r"(a) $\rho = 0.9$")

ax1.set_axis_off()  # Turn off all axis lines, ticks, and labels
ax1.xaxis.pane.fill = False  # Remove fill from x-axis pane
ax1.yaxis.pane.fill = False  # Remove fill from y-axis pane
ax1.zaxis.pane.fill = False  # Remove fill from z-axis pane
ax1.xaxis.pane.set_edgecolor("none")  # Remove edge color from x-axis pane
ax1.yaxis.pane.set_edgecolor("none")  # Remove edge color from y-axis pane
ax1.zaxis.pane.set_edgecolor("none")  # Remove edge color from z-axis pane
ax1.set_facecolor("none")  # Set background color to transparent
fig.patch.set_facecolor("none")  # Set figure background to transparent


plt.tight_layout()
plt.savefig(
    Path(savepath) / "Vector" / "bivariate_gaussian.eps",
    format="eps",
    transparent=True,
    bbox_inches="tight",
)
plt.savefig(
    Path(savepath) / "Raster" / "bivariate_gaussian.png",
    format="png",
    transparent=True,
    bbox_inches="tight",
)
plt.show()

# Create the figure
fig = plt.figure(figsize=(5, 5))

# Define components for the mixture of 4 Gaussians
components = [
    # Component 1: centered at (1, 1) with positive correlation
    ((1, 1), np.array([[1.0, 0.7], [0.7, 1.0]])),
    # Component 2: centered at (-1, -1) with negative correlation
    ((-1, -1), np.array([[0.8, -0.5], [-0.5, 0.8]])),
    # Component 3: centered at (0, 2) with no correlation
    ((0, 2), np.array([[0.5, 0], [0, 0.5]])),
    # Component 4: centered at (-2, 1) with moderate negative correlation
    ((-2, 1), np.array([[0.7, -0.3], [-0.3, 0.7]])),
]

# Define weights for the mixture (must sum to 1)
# weights = [0.25, 0.25, 0.25, 0.25]
weights = [0.2, 0.2, 0.2, 0.4]

# Create a single 3D plot for the mixture
ax = fig.add_subplot(1, 1, 1, projection="3d")
plot_gaussian_mixture(ax, components, weights, title="Mixture of 3 Gaussians")

# Remove axes and background for a cleaner look
ax.set_axis_off()  # Turn off all axis lines, ticks, and labels
ax.xaxis.pane.fill = False  # Remove fill from x-axis pane
ax.yaxis.pane.fill = False  # Remove fill from y-axis pane
ax.zaxis.pane.fill = False  # Remove fill from z-axis pane
ax.xaxis.pane.set_edgecolor("none")  # Remove edge color from x-axis pane
ax.yaxis.pane.set_edgecolor("none")  # Remove edge color from y-axis pane
ax.zaxis.pane.set_edgecolor("none")  # Remove edge color from z-axis pane
ax.set_facecolor("none")  # Set background color to transparent
fig.patch.set_facecolor("none")  # Set figure background to transparent

plt.tight_layout()
plt.savefig(
    Path(savepath) / "Vector" / "gaussian_mixture_post.eps",
    format="eps",
    transparent=True,
    bbox_inches="tight",
)
plt.savefig(
    Path(savepath) / "Raster" / "gaussian_mixture_post.png",
    format="png",
    transparent=True,
    bbox_inches="tight",
    dpi=800,
)
plt.show()
```
